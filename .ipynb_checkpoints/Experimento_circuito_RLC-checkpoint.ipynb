 ## Métodos de Runge-Kutta aplicados al modelo del circuito RLC serie.

En este cuaderno de programación se presenta el procedimiento y el código para la obtención de la respueta al escalón de un circuito RLC serie. El circuto se compone de un transformador de 127 a 12 volts con una corriente de salida máxima de 0.25 Amperes, cuya bobina primara es la que te utiliza como la inductancia, la resistencia es un potenciómetro de 5 $K\Omega$ y el capacitor es electrolítico de 1 $\mu F$. El escalón tiene una amplitud de 1.63 Volts generados por medio del convertidor digital-analógico de una placa Arduino Due. Los elementos mencionados se muestran en la figura de abajo.

```python
from __future__ import division, absolute_import, print_function
import serial   # Necesario para la comunicación serrial
import array    # una variable arreglo para almacenar los datos recibidos
import matplotlib.pyplot as plt
import matplotlib.pylab as plb
import numpy as np
from optparse import OptionParser   # para desempaquetar o armar las opciones
import time   # que permitan armar archivos  sus nombres.
from scipy import signal
plt.rcParams.update({'font.size': 11})
```

Las siguientes líneas son necesarias para establecer el número de muestras a adquirir, deben ser potencias de 2 para efectos del procesamiento digital.

```python
samplingRate =5000      # Es mejor si el cociente 42000000/samplingRate es un entero                   
NFFT = 2**12    #Debe ser una potencia de 2
nSamples = 2**10                      # debe ser una potencia de 2 debe ser 512 o mayor.

a = array.array('H')  # los datos se almacenan aquí
                      # 'H' significa byte sin signo, dos localidades
# Se inicializa el puerto serie.
ser = serial.Serial(port='COM4',
                    baudrate=2000000,
                    parity=serial.PARITY_NONE,
                    stopbits=serial.STOPBITS_ONE,
                    bytesize=serial.EIGHTBITS)
```

Una vez establecido la comunicación USB, se procede a enviar el valor numérico del numero de muestras a tomar y la frecuencia de muestreo. Se hace la adquisión de los datos, al ejecutar las siguientes líneas.

```python
ser.write((str(nSamples) + '\n' + str(samplingRate) + '\n').encode(encoding='ascii',errors='strict'))
# Lectura de datos
# Es necesario transmitir 4 bytes por cada muestra.
# 2 bytes el número aleatorio, 2 bytes de los datos adquiridos, 
# así cada paquete de 512 bytes incluye 128 muestras. 
print("Iniciando adquisición de datos")
if ser.isOpen():
    for i in range(nSamples//128):
        a.fromstring(ser.read(512))

# Se cierra el puerto de comunicaciones.
ser.close()

# Se procesan los datos.
print("Adquisición terminada... Datos disponibles para procesamiento")
```

Es posible almacenar los datos aquiridos en un archivo en formato .txt, el almacenamiento se realiza en las siguientes líneas.

```python
b = np.array(a)  #Se crea un arreglo para almacenar los datos de la conversión analógica-digital.
adc = b[b < 2**13]  
f= open("RLC_lineal_escalon.txt","w+")
i=0
for i in range(len(adc)):
     f.write(str(i*(1/samplingRate)))
     f.write(",")
     f.write(str(round((3.3*adc[i]/4095),4)))
     f.write("\n")
     i=i+1
f.close() 

# Nombre del archivo
filename=str(int(time.time())) + "_" + str(nSamples) + "_" + str(samplingRate)
```

```python
R = 1600;
L =16.0;
C = 1.01*(10.0**(-6));  
t=np.arange(0,nSamples*(1/samplingRate),1/samplingRate)
```

```python
lti = signal.lti([1], [L,R,1/C])
t_final=0.205
t_sim=np.linspace(0,t_final,int(t_final*5000))

t_sim,y = signal.step(lti,[0,0],t_sim)

fig1, (ax1) = plt.subplots(1,sharex=False, sharey=False)
ax1.plot(t,(3.3*(adc)/ 4096))
ax1.plot(t_sim, y*(1.63/C))
ax1.set_title('Voltaje')
ax1.set_xlabel('tiempo [s]')
ax1.set_ylabel('V [volts]')
ax1.set_title('Respuesta al escalón del sistema RLC')
ax1.legend(["Experimental","Simulación"], numpoints=1,loc=0)
ax1.grid()

#print(np.size(t))


fig2, (ax2) = plt.subplots(1,sharex=False, sharey=False)

ax2.magnitude_spectrum(adc-adc.mean(),Fs=samplingRate, color='C1',scale='dB')
ax2.set_xlabel('Frecuencia [Hz]')
ax2.set_ylabel('Magnitud')

w_final=1600
w=(2*3.1416)*np.linspace(0,w_final,int(w_final*500))
w,mag, phase = signal.bode(lti,w)
ax2.plot((1/(2*3.1416)*w),2*mag+250)
ax2.set_xlabel('Frecuencia [Hz]')
ax2.set_ylabel('Magnitud [db]')
ax2.set_xlim(1,1000)
ax2.grid()
ax2.legend(["Experimental","Simulación"], numpoints=1,loc=0)

```
